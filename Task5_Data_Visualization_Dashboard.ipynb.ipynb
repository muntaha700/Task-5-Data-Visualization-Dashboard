```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('coretech_clients_cleaned.csv')

# -------------------------
# 1. Bar Chart - Clients by Service
# -------------------------
plt.figure(figsize=(8,5))
df['service'].value_counts().plot(kind='bar')
plt.title('Clients by Service')
plt.xlabel('Service')
plt.ylabel('Number of Clients')
plt.tight_layout()
plt.savefig('bar_clients_by_service.png')
plt.show()

# -------------------------
# 2. Pie Chart - Lead Source Distribution
# -------------------------
plt.figure(figsize=(7,7))
df['lead_source'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Lead Source Distribution')
plt.ylabel('')
plt.savefig('pie_lead_source.png')
plt.show()

# -------------------------
# 3. Line Chart - Average Project Duration by Service
# -------------------------
service_duration = df.groupby('service')['project_duration_days'].mean()

plt.figure(figsize=(8,5))
plt.plot(
    service_duration.index,
    service_duration.values,
    marker='o'
)
plt.title('Average Project Duration by Service')
plt.xlabel('Service')
plt.ylabel('Average Duration (Days)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('line_project_duration.png')
plt.show()

# -------------------------
# 4. Histogram - Ratings Distribution
# -------------------------
plt.figure(figsize=(8,5))
plt.hist(df['rating'], bins=10)
plt.title('Ratings Distribution')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('histogram_ratings.png')
plt.show()

# -------------------------
# 5. Scatter Plot - Budget Analysis
# -------------------------
plt.figure(figsize=(8,5))
plt.scatter(df['project_budget'], df['rating'])
plt.title('Project Budget vs Rating')
plt.xlabel('Project Budget')
plt.ylabel('Rating')
plt.tight_layout()
plt.savefig('scatter_budget_analysis.png')
plt.show()

print("All charts generated successfully!")
```
